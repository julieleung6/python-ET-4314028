### Removing duplicates from a list

```python
myList = ['a', 'b', 'b', 'c', 'c']
```

```python
myList = set(myList)
myList
```

```python
myList = list(myList)
myList
```

```python
#same as above
myList = ['a', 'b', 'b', 'c', 'c']
myList = list(set(myList))
myList #deduplicated the list
```

### Properties of Sets

- Order doesn't matter
- Declared with curly brackets {}
- All elements are **unique**

NOTE: Sets are randomized, the *order of elements* in your list *may not be the same* coming back out. Because of this, you can't fetch elements by their index like you can with a list.

```python
mySet[0]
```

Set object is not subscriptable, so you can't use the slicing syntax with it. It doesn't make any sense to get the first element in a big random pile. 

In Python, an object is **subscriptable** if it contains elements that can be **accessed by an index**. It contains ordered, accessible sub-elements.

```python
#The int object is not subscriptable. The number, 1 doesn't contain any sub elements.
number = 1
1[0]
```

## Tuples

- Declared with parentheses
- Order matters
- Subscriptable
- Cannot be modify

```python
myTuple = ('a', 'b', 'c')
myTuple
```

```python
myTuple[0]
```

```python
# Tuple doesn't support item assignment
myTuple[0] = 'd'
```

### Why use Tuples

- More efficient than lists
- They don't grow or change
- Store compactly in memory

```python
#Tuples are used by default
def returnsMultipleValues():
    return 1,2,3 

type(returnsMultipleValues())
```

```python
def returnsMultipleValues():
    return (1,2,3)

type(returnsMultipleValues())
```

```python
myTuple = (1,2,3) #don't have to have parentheses but it's preferred!!
```

```python
type(myTuple)
```

### Unpacking values

```python
# setting multiple variables in a line
a, b, c = returnsMultipleValues()
```

```python
print(a)
print(b)
print(c)
```
