## For

```python
myList = [1,2,3,4]
for item in myList:
    print(item)
```

### Pass

- `pass` statement is used as a placeholder, come back and fill it in later.

```python
animalLookup = {
    'a': ['aardvark', 'antelope'],
    'b': ['bear'],
    'c': ['cat'],
    'd': ['dog'],
}

for letter, animals in animalLookup.items():
    pass
```

### Continue

- `continue` statement is for skipping the rest of the loop during each iteration.

```python
for letter, animals in animalLookup.items():
    if len(animals) > 1:
        continue
    print(f'Only one animal! {animals[0]}')
```

### Break

- `break` statement is for stopping partway if we found what we're looking for.

```python
for letter, animals in animalLookup.items():
    if len(animals) > 1:
        print(f'Found {len(animals)} animals: {animals}')
        break
```

### For / Else

- AKA `break/else` statement

**Note:** This `break/else` pattern can also be used with *while* loops.
If you put a `break` in your while loop, you can add an `else` statement at the end of it immediately after that loop. And that `else` only happens if the `break` wasn't thrown.

```python
for number in range(2, 100):
    for factor in range(2, int(number ** 0.5) + 1):
        if number % factor == 0:
            break
    else:
        print(f'{number} is prime!')
        
```

**Note:** The following cell has been modified from what was shown in the video in order to fix a bug. The variable <code>found_factors</code> was incorrectly spelled as <code>found_factor</code>

```python
# the same code as above but with an extra variable, found_factors
# the variable, found_factors, is used to keep track of whether or not a break happened.
for number in range(2, 100):
    found_factors = False
    for factor in range(2, int(number ** 0.5) + 1):
        if number % factor == 0:
            found_factors = True
    if not found_factors: # instead of the else statement above, we have an if not found_factors. They do the same thing
        print(f'{number} is prime!')
```
