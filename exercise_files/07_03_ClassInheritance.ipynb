## Class Inheritance

In Python, a class can extend another class.

The original class is called the **parent class**, and the class extending it is the **child class**.
The *child class* inherits all of the *parent classes'* methods and attributes. This inheritance happens automatically when the *child class* is created.

`super()` is used inside a class to get an instance of the parent class.

Assuming that `rover` is an instance of the `Dog`class, the equivalent of `rover.speak()` is `Dog.speak(rover)`

```python
class Dog: #parent class
    _legs = 4
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(self.name + ' says: Bark!')
    
    def getLegs(self):
        return self._legs


class Chihuahua(Dog): #child class
    pass
```

```python
dog = Chihuahua('Roxy')
dog.speak()
```

We use parentheses after the class name (`class Chihuahua(Dog):`) to indicate that this class, `Chihuahua` is extending the parent `Dog` class.

```python
class Dog: #parent class
    _legs = 4
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(self.name + ' says: Bark!')
    
    def getLegs(self):
        return self._legs


class Chihuahua(Dog): #child class
    def speak(self): #extending this method
        print(f'{self.name} says: Yap yap yap!')
        
    def wagTail(self): #adding new method
        print('Vigorous wagging!')
```

If the child class defines any attributes or methods that conflict with the parent class, the **parent class's methods get overwritten**, and the **child class method is used** instead.

Can also add additional methods to the child class.

```python
dog = Chihuahua('Roxy')
dog.speak()
```

```python
myDog = Dog('Rover')
myDog.speak()
```

```python
dog = Chihuahua('Roxy')
dog.speak()
dog.wagTail()
```

```python
myDog = Dog('Rover')
myDog.speak()
```

This can be extremely useful! For example, there's a class that you really want to use but it just needs a couple tweaks or maybe an extra method. We can even do this with Python's own built-in classes.

### Extending built-in classes

```python
#instantiate a new list
myList = list() #even tho this is lowercase and looks like a function, list() is actually a class 
```

```python
# DONT RUN THIS CODE!!!

class UniqueList(list): #extending the list class to make our own UniqueList class
    def append(self, item): #overriding the append function
        if item in self:
            return
        self.append(item) #this calls the exact same function, append ^^ AND this will lead to INFINITE RECURSION!! 
        # Never ending loop and break our program
        
uniqueList = UniqueList()
uniqueList.append(1)
uniqueList.append(1)
uniqueList.append(2)

print(uniqueList)
```

We want to call append in the parent class, the original `list` class. In this case, we use a function called **super**, which gets the underlying instance of the parent class.

```python
class UniqueList(list):
    def append(self, item):
        if item in self:
            return
        super().append(item) #FIXED!!
        
uniqueList = UniqueList()
uniqueList.append(1)
uniqueList.append(1)
uniqueList.append(2)

print(uniqueList)
```

There is also another really common use case where **super** is used, and that's in the *constructor*.


Let's say we want to add another attribute to our child class instance, we could use `__init__(self)...` The problem is that we are completely overwriting the constructor and the parent class now, where it may have some really important stuff that it needs to initialize. We can solve this by using `super` again. This makes sure that the parent constructor gets called first, and then we add 
`self.someProperty = 'Unique List!'`

```python
class UniqueList(list):
    
    def __init__(self):
        super().__init__() # this ensures the parent constructor gets called first
        self.someProperty = 'Unique List!'
        

    def append(self, item):
        if item in self:
            return
        super().append(item)
        
uniqueList = UniqueList()
uniqueList.append(1)
uniqueList.append(1)
uniqueList.append(2)

print(uniqueList.someProperty) #instantiating this
```
