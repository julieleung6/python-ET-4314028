## JSON

### Loading JSON

```python
jsonString = '{"a": "apple", "b": "bear", "c": "cat"}'
```

**NOTE**: JSON is not Python!\
This JSON formatted string looks like a Python dictionary, but *it's not*.

```python
jsonString = '{"a": "apple", "b": "bear", "c": "cat"}'
json.loads(jsonString)
```

This is a string that just happens to be in the JSON format.\
In order to turn it into a dictionary, we need to import the JSON module, `import json` at the top of the notebook. Then we are going to use a method called `json.loads` and pass in the string, `jsonString`.

```python
{"a": "apple", "b": "bear", "c": "cat"}
```

Now, we have a Python dictionary. I can add a trailing comma and that's fine.

```python
{"a": "apple", "b": "bear", "c": "cat",}
```

But if we add a trailing comma to JSON string, we get a `JSONDecodeError`.

```python
jsonString = '{"a": "apple", "b": "bear", "c": "cat",}'
json.loads(jsonString)
```

It's very common if you are working with JSON from a potentially untrustworthy source to surround this `json.loads()` with a `try/except`.

```python
jsonString = '{"a": "apple", "b": "bear", "c": "cat",}'
try:
    json.loads(jsonString)
except JSONDecodeError:
    print('Could not parse JSON!')
```

### Dumping JSON (dumping a Python dictionary into a JSON string)

- use `json.dumps()` method

```python
pythonDict = {'a': 'apple', 'b': 'bear', 'c': 'cat',}
json.dumps(pythonDict)
```

You usually wouldn't add any exception handling in this case because if you have a valid Python dictionary, there's not a lot that could go wrong when you're formatting it as a JSON string.\
However, there is one exception to this rule where an exception could be thrown.

### Custom JSON Decoders

```python
class Animal:
    def __init__(self, name):
        self.name = name
    
pythonDict = {'a': Animal('aardvark'), 'b': Animal('bear'), 'c': Animal('cat'),}
json.dumps(pythonDict)
```

You see we get a `TypeError: Object of type Animal is not JSON serializable`.\
The JSON module has no idea how to handle this `Animal` class. It doesn't know what the JSON equivalent should be.\
So, here we override the default JSON encoder that it's using with a **JSON encoder** of our own.

```python
class Animal:
    def __init__(self, name):
        self.name = name

class AnimalEncoder(JSONEncoder):
    def default(self, o): #o is the object that is being passed in here that needs to be decoded into JSON.
        if type(o) == Animal:
            return o.name
        return super().default(o) #if its not animal,pass it off to parent version of encoder by using super, then call that default method.
    
pythonDict = {'a': Animal('aardvark'), 'b': Animal('bear'), 'c': Animal('cat'),}
json.dumps(pythonDict, cls=AnimalEncoder) #need to let json.dumps know that it needs to use AnimalEncoder by passing the cls keyword argument
```

Again, generally, you don't need to surround `json.dumps()` in a `try/except` statement because you are usually creating the dictionaries that you pass into `json.dumps()` in your own code, and you should appropriately handle any data types that you are passing in there.\
But if you get any crazy JSON formatted files from unknown sources, it's best to beware!!!
