## List Comprehensions

- allows you to essentially make a *for loop* in one line while returning a copy of the list that you're iterating over.
- allows you to filter or call functions on every item or of a list.

```python
myList = [1,2,3,4,5]
[2*item for item in myList]
```

### List comprehensions with filters

```python
myList = list(range(100))
filteredList = [item for item in myList if item % 10 == 0]
filteredList
```

```python
filteredList = [item for item in myList if item % 10 < 3]
print(filteredList)
```

### List comprehensions with functions

```python
myString = 'My name is Ryan Mitchell. I live in Boston'
myString.split('.')
```

```python
myString.split() #will split on spaces by default
```

```python
#remove the period and then strings to lowercase
def cleanWord(word):
    return word.replace('.', '').lower() # chaining functions (calling one function after the other in a single line)

[cleanWord(word) for word in myString.split()]
```

```python
[cleanWord(word) for word in myString.split() if len(cleanWord(word)) < 3]
```

### Nested list comprehensions

```python
# a list comprehension within a list comprehension
# this is splitting the original text into sentences then performing an inner list comprehension on each sentence
# so we have a list of lists. grouped the clean words by the sentence that they appeared in
[[cleanWord(word) for word in sentence.split()] for sentence in myString.split('.')]
```

```python
myString = 'My name is Ryan Mitchell. I live in Boston'

for sentence in myString.split('.'):
    for word in sentence.split():
        print(cleanWord(word))
```

```python
myString = 'My name is Ryan Mitchell. I live in Boston'

combine_lists = []

for sentence in myString.split('.'):
    sent_list = []
    for word in sentence.split():
        print(cleanWord(word))
        sent_list.append(cleanWord(word))
    print(sent_list)
    combine_lists.append(sent_list)
    
combine_lists
```
