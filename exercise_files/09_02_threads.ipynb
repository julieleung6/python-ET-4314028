```python
import threading
import time
```

## Threads

- Threads come in handy when your code is just sitting around doing nothing while waiting for that data to come back.
- You can do that waiting in parallel rather than one at a time.

Threads share memory and can modify the same object.

```python
def longSquare(num):
    time.sleep(1)
    return num**2

[longSquare(n) for n in range(0, 5)]
```

### Creating two threads

```python
t1 = threading.Thread(target=longSquare, args=(1,)) #(1,) bc we want tuples
t2 = threading.Thread(target=longSquare, args=(2,)) 

t1.start()
t2.start()

t1.join()
t2.join()
```

The return value is nowhere in these threads. There is no way to get the output of this function directly.

```python
#creating a results dictionary
def longSquare(num, results):
    time.sleep(1)
    results[num] = num**2

results = {}
t1 = threading.Thread(target=longSquare, args=(1,results))
t2 = threading.Thread(target=longSquare, args=(2,results))

t1.start()
t2.start()

t1.join()
t2.join()

print(results)
```

```python
#putting the threads into a list

def longSquare(num, results):
    time.sleep(1)
    results[num] = num**2

results = {}
threads = [threading.Thread(target=longSquare, args=(n, results)) for n in range(0, 100)]
[t.start() for t in threads]
[t.join() for t in threads]

print(results)
```

### Threading

- Emulates parallel computing
- Useful when your programs have periods of downtime

### Multi-threading 

- DOESN'T always decrease the runtime of a program.
- But when the program has periods of "waiting" and doing nothing, it will decrease the runtime of a program.
