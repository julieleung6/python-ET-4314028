## Errors and Exceptions

```python
1/0
```

Working with Python, you'll see that sometimes things are called **errors** and sometimes they're called **exceptions**. If you read the official Python documentation, there are so many exceptions to the rule so it's not really worth worrying too much about. They all work the same anyway. Errors and exceptions are the same thing (controversial opinion).

All Python errors and exceptions ultimately extend to a class called the **base exception**.
So, `division by zero` extends `ArithmeticError` which extends the exception which extends `BaseException`. And `BaseException` is the thing that gives us really useful and powerful properties of exceptions, which is to halt the execution of code and give you some information about how and why that execution was halted.

```python
def causeError():
    return 1/0

causeError()
```

We can see where the function was called originally, then we can see where in that function that we called the error happened. And this is called a **stack trace** (often called a traceback).

If you think about these nested calls like a stack of operations, it provides a trace through the stack that we can use to debug our program.

```python
#making the stack trace even longer
def causeError():
    return 1/0

def callCauseError():
    return causeError()

callCauseError()
```

Exceptions are just classes.

### Try / Except

We can catch an exception using a `Try/Except` statement, and we can get an instance of that exception that was raised.

```python
try:
    1/0
except Exception as e: #e is going to be our variable. Actually that instance of the exception that was raised.
    print(type(e))
```

Exceptions when used and handled correctly are almost like a secondary layer of code underlying all the existing and more obvious code.
