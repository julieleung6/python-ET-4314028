## Files

When working with files in Python, we are working a little more directly with the operating system so there are some things that we need to manage.

One of those things is whether we are simply *reading the contents* of the file or whether we intend to *make changes* to it, *writing it*. This is because it causes problems if two applications are making changes to the same file at the same time. So the operating system needs to know who's doing what.

### Reading Files

```python
f = open('10_01_file.txt', 'r') #read mode
print(f)

#output: we get a file object
```

### `f.readline()`


`f.readline()` reads the lines of the file, `f` one at a time.\
When we run the code again, we get a different line each time.\
This file object contains a sort of bookmark of which lines of the file it's already read.

```python
f.readline()
```

### `f.readlines()`


`f.readlines()` will give us all of the lines of the file.\
This gets all of the lines of the file that haven't been read already and then puts them into a list of strings.

```python
f.readlines()
```

```python
f = open('10_01_file.txt', 'r')
for line in f.readlines():
    print(line)
```

**Notice**: These lines are all double-spaced and that's because each line of the file has a new line character on the end there.\
We can fix this by stripping out any leading or trailing white space, including new lines using `.strip()` on each line.

```python
f = open('10_01_file.txt', 'r')
for line in f.readlines():
    print(line.strip())
```

### Writing Files

A new file, `10_01_output.txt`, which doesn't exist yet will be created after we run the code below.

```python
f = open('10_01_output.txt', 'w') #write mode
print(f)
```

```python
#created new file
#now let's write a couple of lines in it
f.write('Line 1')
f.write('Line 2')

#Output: Opened the file but there was nothing there
```

Writing to files is an *expensive* operation and Python tries to make file writing efficient by putting all of the data you are writing to the file in a buffer.\
It only writes to the file when that buffer gets full or when you close the file.\
So, if we close the file, we can do that with `f.close()` and then run that.

```python
f.close()
```

Output: `Line 1Line 2`\
It didn't print a new line character between the lines.\
We have to add that ourselves when we are writing files.

```python
#let's add the new line character
# open this file again, write those lines, then close it
f.write('Line 1\n')
f.write('Line 2\n')

#reopening the file. Now, we see our two lines.
```

When we open a file in `write mode` and start writing to it, it's actually more like a create mode.\
Python will **overwrite any existing data in that file**.\
We can fix that by opening the file in `append mode` instead.

### Appending Files

```python
f = open('10_01_output.txt', 'a') #append mode
f.write('Line 3\n')
f.write('Line 4\n')
f.close()
```

It's very important that `f.close()` gets called.\
It basically releases the file, telling the operating system we are done writing to it.

These files will eventually get closed. There's a process in Python that cleans up old variables that aren't being used anymore and will close them. However, that behavior is unpredictable and you might get some kind of strange behavior if you're not managing the closing of these files yourself.

**Best Practice**: To close them when you are done with them. The most common way to do that is with the `with` statement.

```python
with open('10_01_output.txt', 'a') as f:
    f.write('some stuff\n')
    f.write('some other stuff\n')
#As soon we outdent and leave this code block, this file gets closed!!! 

print(f)
```

Notice: We still have access to this variable, `f`. So if we run this `print(f)`, it will write that `f` still exists.

```python
f.write('PS. I forgot some stuff')
```

We got an `I/0 error` or an input/output error because this file is already closed for writing.
