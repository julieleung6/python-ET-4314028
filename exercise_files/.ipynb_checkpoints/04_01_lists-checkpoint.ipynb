## Lists

### List Slicing

```python
myList = [1,2,3,4,5]
myList[3:]
```

```python
myList[0:]
```

```python
#start value, end value, and the step size of 2
myList[0:6:2]
```

```python
myList[0:6:3]
```

```python
myList[:]
```

```python
myList[::2]
```

### range()

- It has an order
- Immuntable
- Used in for loops

```python
for i in range(100):
    print(i)
```

```python
#take the range and convert it to a list
myList = list(range(100))
```

```python
myList[::15]
```

```python
myList[::10]
```

```python
#stepping backwards
myList[::-10]
```

### Modifying Lists

```python
myList = [1,2,3,4]
myList.append(5)
print(myList)
```

```python
#insert alows you to insert a value at any position
#.insert(index number, desired value)
myList.insert(3, 'a new value')
print(myList)
```

```python
myList.remove('a new value') #dont need to pass the index number, only the value

#will get an error when you remove an item that isn't on the list
```

```python
myList
```

```python
myList.pop() #removes the last item off the list
```

```python
myList
```

```python
while len(myList):
    print(myList.pop())
```

```python
myList
```

```python
a = [1,2,3,4,5]
b = a
a.append(6)
print(b)
#modifying the list, a also modifies the list, b
```

```python
#You can see the difference between setting a variable to another variable VS.
# Making a copy of it and setting it to the value of that variable
a = [1,2,3,4,5]
b = a.copy()
a.append(6)
print(a)
print(b)
```

```python
for i in range(20):
    print(i)
```

**Lists** are one of the **most fundamental** and **useful** data structures in Python.
