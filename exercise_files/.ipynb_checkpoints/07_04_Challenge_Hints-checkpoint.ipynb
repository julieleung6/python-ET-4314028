**Hint 2:** Triangles are small at the top and big at the bottom! (Huge help, I know.) 

But think about it. How can you use the variable i to print out something small at the top and big at the bottom? Or, heck, make an upside-down triangle! Print out something big at the top and small at the bottom. 

```python
# Python code​​​​​​‌‌‌‌​‌​‌‌​‌‌​​‌​‌​​‌‌‌‌​‌ below
class Shape:
	width = 5 #default width value
	height = 5 #default height value
	printChar = '#' #this is the character we use to print out the body of our ASCII art shapes.

	def printRow(self, i):
		raise NotImplementedError("Will be implemented by children extending this class")

	def print(self): #print method
		for i in range(self.height): #iterates through the height, printing out one row at a time.
			self.printRow(i) #calling printRow with height index, where 0 is the top row and 4 (using default height value), is bottom row

class Square(Shape):
	def printRow(self, i): #implement printRow here and each row is exactly the same.
		print(self.printChar * self.width) #Squares are the same at the top as they are on the bottom
        #It doesnt even use the height index that gets passed in at all.

class Triangle(Shape):
    def printRow(self, i):
        print(self.printChar * (i + 1)) #multiple printChar by height index + 1
```

```python
class Triangle(Shape): #symmetrical triangle
    height = 5
    width = 2 * height #need it to be wider
    #because for each row to stack on top of subsequent larger rows in a symmetrical way, the # of characters in each row needs to be odd.
    def printRow(self, i):
        triangleWidth = i * 2 + 1 #this will guarantee the width of our triangle be an odd number
        #need some initial padding space before we start printing the row. The padding space is going to get smaller as we go down.
        #we can calculate the pading as the width of the entire figure, which is 2 * height - triangleWidth
        padding = int((self.width - triangleWidth)/2) #dividing by 2 so we have equal padding on both sides
        print(" "* padding + self.printChar * triangleWidth) #dont need to print out the padding on the right hand side bc that is implied
```

```python
symmetrical_triangle = Triangle()
```

```python
symmetrical_triangle.print()
```
