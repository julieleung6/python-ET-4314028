### Text processing in Python

```python
text = '''
Beautiful is better than ugly.
Explicit is better than implicit.
Simple is better than complex.
Complex is better than complicated.
Flat is better than nested.
Sparse is better than dense.
Readability counts.
Special cases aren't special enough to break the rules.
Although practicality beats purity.
Errors should never pass silently.
Unless explicitly silenced.
In the face of ambiguity, refuse the temptation to guess.
There should be one-- and preferably only one --obvious way to do it.
Although that way may not be obvious at first unless you're Dutch.
Now is better than never.
Although never is often better than *right* now.
If the implementation is hard to explain, it's a bad idea.
If the implementation is easy to explain, it may be a good idea.
Namespaces are one honking great idea -- let's do more of those!
'''
```

```python
def lowercase(text):
    return text.lower()

def removePunctuation(text):
    punctuations = ['.', '-', ',', '*']
    for punctuation in punctuations:
        text = text.replace(punctuation, '')
    return text

def removeNewlines(text):
    text = text.replace('\n', ' ')
    return text

def removeShortWords(text):
    return ' '.join([word for word in text.split() if len(word) > 3])

def removeLongWords(text):
    return ' '.join([word for word in text.split() if len(word) < 6])
```

```python
#can mix and match the order I perform these functions in and which functions I apply by calling them in a list
processingFunctions = [lowercase, removePunctuation, removeNewlines, removeLongWords]

for func in processingFunctions:
    text = func(text)

print(text)
```

### Lambda functions

- Convenient way to write "mini functions" as values

One last way to represent a function that you might use and encounter. 

So, not every piece of data needs a variable name, right? Like I can just type `5` and that kind of exists.

You can do the same with functions as well and they are called **lambda functions**. 

```python
5
```

```python
2 + 3
```

```python
#You can define a small function without a variable name, lambda x, and then a colon, x + 3
#and we can call it then with 5 and that returns 8
(lambda x: x + 3)(5)
```

You type `lambda` followed by your parameter names. In this case, just that single parameter `x`, then a colon, and then a *one* line function.

**No multi-line functions with lambda functions**

Don't need to type `return`. The return is automatically assumed. There's an implied return, as we call it.

This can come in handy in some Python functions that take in a function as their argument.

```python
myList = [5,4,3,2]
sorted(myList)
```

For instance, for sorting a list of values you might use the function sorted. Great, it sorted my list of values.

But if the things you're sorting don't have an obvious numeric value, you can pass in a function that takes an item in the list and returns the value that Python should be using to sort it.

```python
myList = [{'num': 3}, {'num': 2}, {'num': 1}]
sorted(myList)
```

You can see that we can't sort it. But, if we pass in this `key=` parameter, which is a function, `lambda x`, `x` is going to be our item in here, and then return `x` number, and now it can sort it.

```python
myList = [{'num': 3}, {'num': 2}, {'num': 1}]
sorted(myList, key=lambda x: x['num']) #For each item, look at the value inside 'num', and use that to sort.
#x is not the whole list — it is one dictionary (one item) from the list at a time.
#x['num'] = the number inside that dictionary
```

When sorting dictionaries, it's usually best if they all share the **same key**.

```python
myList = [
    {'num': 3},
    {'value': 2},
    {'num': 1}
]

sorted(myList, key=lambda x: x['num'])
```

```python
myList = [
    {'num': 3},
    {'value': 2},
    {'num': 1}
]

sorted(myList, key=lambda x: x.get('num', 0))
```

```python
car = {"brand": "Ford", "model": "Mustang"}
car["year"] = 1964 # Adds a new item
car["year"] = 2020 # Modifies the existing item
```

```python
for key, value in car.items():
    print(f"Key: {key}, Value: {value}")
```
