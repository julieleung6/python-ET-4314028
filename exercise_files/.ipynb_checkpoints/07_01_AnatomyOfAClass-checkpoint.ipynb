## Class

The `self` variable is used inside a class because it is the conventional variable name to refer to the class instance.

### Instance Attributes

```python
class Dog: #dog class
    def __init__(self, name): #two instance attributes: name and legs.
        self.name = name
        self.legs = 4
    
    def speak(self):
        print(self.name + ' says: Bark!')

myDog = Dog('Rover') #a new instance
print(myDog.name)
print(myDog.legs)
```

We call them instance attributes because they are attributes that each instance of the dog class has.

```python
Dog.legs
```

Notice that even though this value (`self.legs = 4`) is hard coded in the dog initialization function, we can't see what the value is directly. 

So, if we say `Dog.legs`, we get an error. But we can't change the value of legs. So what if we decide that having four legs is an intrinsic property of being a dog?

### Static Attributes

Let's make legs an attribute of the class itself. We can do that by moving it outside of the initialization function of the constructor.

```python
class Dog:
    legs = 4 #define it here
    def __init__(self, name):#instead of being inside here
        self.name = name 
    
    def speak(self):
        print(self.name + ' says: Bark!')

myDog = Dog('Rover')
print(myDog.name)
print(myDog.legs)
```

```python
Dog.legs
```

**Static attributes or static variables** are unchanging with each instance. They're not dynamic, they're static.

Traditionally, static variables are used to hold constants and fundamentals business logic and that sort of thing. But be careful. Because just like any variable, you **can reset it to the class**.

You can set the value of legs to something else.

```python
Dog.legs = 3
```

```python
myDog = Dog('Rover')
print(myDog.name)
print(myDog.legs)
```

The convention to prevent this sort of thing is to add an *underscore* before the variable name.

Note: This doesn't actually stop anyone from messing with the variable, but the underscore is just an indicator or a warning, mess with this at your own risk because you could break things by changing the value.

The *underscore* also has another connotation, that the user shouldn't rely on or even reference the values directly. These private variables are implementation details, subject to change without notice, don't look at them.

In this case, we might use what's called a **getter** function, or a **get** method, or a getter method for the legs.

```python
class Dog:
    _legs = 4
    def __init__(self, name):
        self.name = name
        
    def getLegs(self): #dont need to pass the self attribute here
        return self._legs
    
    def speak(self):
        print(self.name + ' says: Bark!')

myDog = Dog('Rover')
print(myDog.name)
print(myDog.getLegs())
```

We don't need to pass the `self` attribute into the `getter` function.

Note: `self` is an object instance that's literally the same instance we're calling the function on.

```python
class Dog:
    _legs = 4
    def __init__(self, name):
        self.name = name
        
    def getLegs():
        return Dog._legs #the problem is that now we cant call this method. it'll throw an error
    
    def speak(self):
        print(self.name + ' says: Bark!')

myDog = Dog('Rover')
print(myDog.name)
print(myDog.getLegs())
```

We have an error because when we call a method on a class instance, like `myDog.getLegs()`, the class instance, `myDog`, gets passed in as the first value automatically, and saying 0 positional arguments but 1 was given. And that's because we're passing this invisible value in there.

```python
class Dog:
    _legs = 4
    def __init__(self, name):
        self.name = name
        
    def getLegs():
        return Dog._legs #the problem is that now we cant call this method. it'll throw an error
    
    def speak(self):
        print(self.name + ' says: Bark!')

myDog = Dog('Rover')
print(myDog.name)
print(Dog.getLegs())
```

We could still use the function like this, `Dog.getLegs()`, but that's just not very intuitive. It looks odd. 

In my opinion, the best way to do this is:\
`def getLegs(self):`\
`return self._legs`\
and just call it in the traditional way.

```python
class Dog:
    _legs = 4
    def __init__(self, name):
        self.name = name
        
    def getLegs(self):
        return self._legs
    
    def speak(self):
        print(self.name + ' says: Bark!')

myDog = Dog('Rover')
print(myDog.name)
print(myDog.getLegs())
```

Note: Classes have their own variable scope rules that are very similar to the variable scope rules with functions.

```python
myDog = Dog('Rover')
myDog._legs = 3 #changing the instance variable, NOT the class variable
print(myDog.name)
print(myDog.getLegs())
print(Dog._legs) # the class variable remains the same
```
