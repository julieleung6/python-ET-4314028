```python
import math
```

## Functions

```python
def performOperation(num1, num2, operation):
    if operation == 'sum':
        return num1 + num2
    if operation == 'multiply':
        return num1 * num2
    
performOperation(2, 3, 'sum')
```

We want to add the parameters together, but we don't want to have to write 'sum' all the time. We can provide a default value using what's called **name parameters** or **keyword arguments**.

### Named Parameters

```python
def performOperation(num1, num2, operation='sum'): #providing a default value for the operation parameter
    if operation == 'sum':
        return num1 + num2
    if operation == 'multiply':
        return num1 * num2
    
performOperation(2, 3)
```

```python
performOperation(2, 3, 'multiply')
```

You don't have to say `operation='multiply'` when you're calling this function. You can just pass multiply in as the third parameter. But, if you have a function with lots and lots of these optional keyword parameters, you don't have to worry about what order you're passing everything in. So sometimes it's easier, more readable to say `operation='multiply'`.

For example, we can add another keyword argument to our function here. A new `message`. As long as we are defining which argument is which.

```python
def performOperation(num1, num2, operation='sum', message='Default message'):
    print(message)
    if operation == 'sum':
        return num1 + num2
    if operation == 'multiply':
        return num1 * num2
    
performOperation(2, 3, message='A new message!', operation='multiply')
```

### Rule
- Keyword arguments **must come after** positional arguments. (In this function, it's the 2 and 3) 
    - The first two arguments, the order matters very much. They're not optional.
- Afterwards, keyword arguments can be in any order.
    -  There optional arguments are nice, but there's still this fundamental limitation or functional limitation on what we can do here.
- We need to anticipate **ALL** the variables the user might be passing in.
    - What if we want to allow them to pass in any number of variables they want?

### *args

```python
def performOperation(args):
    print(args)

performOperation(1,2,3)
```

The way I written this function, I'm telling it that it's supposted to be expecting a *single* variable called `args`, but there are three of them I'm passing in. So, there's a mismatch aka `TypeError`!

The asterisk on the `args` tells Python, the variable name isn't literally args. This is a reference, a pointer to the stuff that's being passed in.

Technically, you don't have to use the word `args` here as this variable name, but you should use args because it's Python standard.

```python
def performOperation(*args):
    print(args)
    
performOperation(1,2,3)
```

When we call this, we get a tuple. But, this only works for *positional arguments*, not keyword arguments.

```python
performOperation(1,2,3, operation='sum')
```

To handle keyword arguments, we need to do this differently with something called `kwargs`, which stands for keyword arguments.

### **kwargs

```python
def performOperation(*args, **kwargs):
    print(args)
    print(kwargs)
performOperation(1,2,3, operation='sum')
```

```python
performOperation(3,7, message = "hi", ok = "what", no = 0)
```

Note: The keyword arguments are a dictionary instead of a tuple. 
- That is because keyword arguments can have keys and values. They can also be passed in any order so you need a dictionary to reference them.

```python
def performOperation(*args, operation='sum'):
    if operation == 'sum':
        return sum(args)
    if operation == 'multiply':
        return math.prod(args)
    
performOperation(1,2,3,6, 7, 8, operation='sum')
```
