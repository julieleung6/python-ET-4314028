```python
from multiprocess import Process
import time
import threading
```

## Processes

- processes can contain multiple threads.

```python
def longSquare(num, results):
    time.sleep(1)
    results[num] = num**2

results = {}
p1 = Process(target=longSquare, args=(1,results))
p2 = Process(target=longSquare, args=(2,results))

p1.start()
p2.start()

p1.join()
p2.join()

print(results)
```

We see there aren't any results. 

Remember, **processes don't share memory.** They get a copy of this dictionary in their own separate memory space, and we have no way of accessing it except if they record it somewhere like a file system or a database.


One thing we can do is print the computed value from within the function itself.

```python
def longSquare(num, results):
    time.sleep(1)
    print(num**2)
    print('Finished computing')

results = {}
p1 = Process(target=longSquare, args=(1,results))
p2 = Process(target=longSquare, args=(2,results))

p1.start()
p2.start()

p1.join()
p2.join()

print(results)
```

```python
#putting the processes into a list
def longSquare(num,results):
    time.sleep(1)
    print(num**2)
    print('Finished computing!')

results = {}
processes = [Process(target=longSquare, args=(n,results)) for n in range(0, 10)]
[p.start() for p in processes]
[p.join() for p in processes]
```

This output looks a bit funky. New lines aren't where you expected them to be. There's overlap between the separate function calls.

```python
results = {}
threads = [threading.Thread(target=longSquare, args=(n, results)) for n in range(0, 10)]
[t.start() for t in threads]
[t.join() for t in threads]
print(results)
```

We usually talk about threads and processes as computing things in parallel. And with processes, that's true. Modern computers have multiple processors and you are literally asking for multiple processor to process your tasks in parallel. 

However, with threads, that same processor will execute a statement from thread A, thread B, thread C, then thread A again, and it basically will go to work on a different thread if one of them is hanging around waiting for something for whatever reason, like a `time.sleep`.
