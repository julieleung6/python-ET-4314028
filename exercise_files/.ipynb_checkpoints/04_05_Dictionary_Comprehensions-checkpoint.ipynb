## Dictionary Comprehensions

```python
animalList = [('a', 'aardvark'), ('b', 'bear'), ('c', 'cat'), ('d', 'dog')]
animals = {item[0]: item[1] for item in animalList} #define both the key/value instead of just a single value like in list comprehensions.
animals
```

```python
animals = {key: value for key, value in animalList}
animals
```

Python allows you to **unpack values** into multiple variables as long as the number of variables you're assigning the values to matches the elements to the data structure.

### dictionary to a list

```python
# items returns a dict_items object containing a list of key-value pairs
animals.items()
```

```python
list(animals.items())
```

```python
[{'letter': key, 'name': value} for key, value in animals.items()]
```

```python
result = []
for key, value in animals.items():
    result.append({'letter': key, 'name': value})
result
```
