## If statements with "FizzBuzz"

```python
# If the number is divisible by 3, print Fizz
# If the number is divisible by 5, print Buzz
# If the number is divisible by 15, print FizzBuzz

# Output should look like this - 1, 2, Fizz, 4, Buzz, Fizz, 7, 8, Fizz, Buzz, 11, Fizz, 13, FizzBuzz, 16, etc.

for n in range(1, 101):
    if n % 15 == 0:
        print('FizzBuzz')
    else:
        if n % 3 == 0:
            print('Fizz')
        else:
            if n % 5 == 0:
                print('Buzz')
            else:
                print(n)
```

### Elif statements

- The **elif** statement must always be proceeded by an **if** statement
- The **else** statement is *optional*

The rule is you write a single **if** statement, any number of **elif** statements and then, optionally, a single **else** statement at the very end to provide some default value if nothing above it matches.

```python
for n in range(1, 101):
    if n % 15 == 0:
        print('FizzBuzz')
    elif n % 3 == 0:
        print('Fizz')
    elif n % 5 == 0:
        print('Buzz')
    else:
        print(n)

    # #can also have another if statement. It will start a NEW block of if/elif/else statements
    # if n % 2 == 0:
    #     print('It is even!')
```

**NOTE:** One issue with *if/else* statements si they can often drag on for TOO many lines!

### Single Line if statements

```python
n = 5
print('Fizz' if n % 3 == 0 else n)
```

```python
# Can store it (^the code above^) in a variable
# In programming, this is known as a ternary operator
fizzBuzz = 'Fizz' if n % 3 == 0 else n
```

A **ternary operator** takes in some Boolean condition (ex: `n % 3 == 0`), evaluates it and returns *one* value if the condition is `true` and another value if the condition is `false`. 

```python
# can string multiple if/else statements
'Fizz' if n % 3 == 0 else 'Buzz' if n % 5 == 0 else n
```

```python
'FizzBuzz' if n % 15 == 0 else 'Fizz' if n % 3 == 0 else 'Buzz' if n % 5 == 0 else n
```

```python
# can even put into a list
['FizzBuzz' if n % 15 == 0 else 'Fizz' if n % 3 == 0 else 'Buzz' if n % 5 == 0 else n for n in range(1, 101)]
```
