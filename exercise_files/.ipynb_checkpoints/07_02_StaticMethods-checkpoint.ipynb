## Static and Instance Methods

```python
class WordSet: #going to contain a set of words
    def __init__(self):
        self.words = set() #equal to an empty set. this set starts out empty, can add to it

    def addText(self, text): #creating a method
        text = self.cleanText(text)
        for word in text.split():
            self.words.add(word)

    def cleanText(self, text): #has `self` instance passed in but it never actually gets used. `self` is a completely unused variable.
        text = text.replace('!', '').replace('.', '').replace(',', '').replace('\'', '') #chaining functions
        return text.lower() #make everything lowercase after stripping out the punctuation

        
wordSet = WordSet() #pass in big blocks of text

wordSet.addText('Hi, I\'m Julie! Here is a sentence I want to add!')
wordSet.addText('Here is another sentence I want to add.')

print(wordSet.words)
```

`def cleanText(self, text):` The "self" instance passed here never actually gets used. `self` is a completely unused variable. 


If we try to remove it, we will get an error because we are still passing the `self` instance into the method, but it only takes a single parameter now.

```python
class WordSet:
    def __init__(self):
        self.words = set()
        
    def addText(self, text):
        text = self.cleanText(text)
        for word in text.split():
            self.words.add(word)

    def cleanText(text):
        text = text.replace('!', '').replace('.', '').replace(',', '').replace('\'', '') 
        return text.lower()

        
wordSet = WordSet()

wordSet.addText('Hi, I\'m Julie! Here is a sentence I want to add!')
wordSet.addText('Here is another sentence I want to add.')

print(wordSet.words)
```

```python
class WordSet:
    def __init__(self):
        self.words = set()
        
    def addText(self, text):
        text = cleanText(text) #removing `self` here BUT we got another error!!!
        for word in text.split():
            self.words.add(word)

    def cleanText(text):
        text = text.replace('!', '').replace('.', '').replace(',', '').replace('\'', '') 
        return text.lower()

        
wordSet = WordSet()

wordSet.addText('Hi, I\'m Julie! Here is a sentence I want to add!')
wordSet.addText('Here is another sentence I want to add.')

print(wordSet.words)
```

We got another error because it had no idea where to look for the function, `cleanText`. It says that it's not defined. It's expecting it to be defined outside of the class, but instead, it's defined in the `WordSet` class.

```python
class WordSet:
    def __init__(self):
        self.words = set()
        
    def addText(self, text):
        text = WordSet.cleanText(text) #fixing it by adding `WordSet`
        for word in text.split():
            self.words.add(word)

    def cleanText(text):
        text = text.replace('!', '').replace('.', '').replace(',', '').replace('\'', '') 
        return text.lower()

        
wordSet = WordSet()

wordSet.addText('Hi, I\'m Julie! Here is a sentence I want to add!')
wordSet.addText('Here is another sentence I want to add.')

print(wordSet.words)
```

```python
class WordSet:
    def __init__(self):
        self.words = set()
        
    def addText(self, text):
        text = WordSet.cleanText(text)
        for word in text.split():
            self.words.add(word)
            
        
    def cleanText(text):
        # chaining functions
        text = text.replace('!', '').replace('.', '').replace(',', '').replace('\'', '')
        return text.lower()
    
        
wordSet = WordSet()

wordSet.addText('Hi, I\'m Ryan! Here is a sentence I want to add!')
wordSet.addText('Here is another sentence I want to add.')

print(wordSet.words)
```

This method, `cleanText` is called a **static method**.

By removing the `self`, the class instance from the parameters that get passed into this method, we've said this method doesn't belong to any particular class instance. It's actually part of the `WordSet` class definition itself.

**Static methods** are used to hold constants, unchanging variables, fundamental business logic.

There are **static attributes**, we saw an example of this previously with the `legs` attribute on the `Dog` class. That's a static attribute because it's part of the class definition rather than being associated with a particular class instance.

Note: Methods like `addText` are **instance methods**. They're methods that belong to a particular instance of the class.


### Static Variable

```python
class WordSet:
    replacePuncs = ['!', '.', ',', '\''] #static variable
    def __init__(self):
        self.words = set()
        
    def addText(self, text): #instance method
        text = WordSet.cleanText(text)
        for word in text.split():
            self.words.add(word)
            
        
    def cleanText(text): #static method
        # chaining functions
        for punc in WordSet.replacePuncs:
            text = text.replace(punc, '')
        return text.lower()
    
        
wordSet = WordSet()

wordSet.addText('Hi, I\'m Ryan! Here is a sentence I want to add!')
wordSet.addText('Here is another sentence I want to add.')

print(wordSet.words)
```

With **static variables**, we have the option to use the class name, `WordSet` or the class instance `self` (if we passed `self` as a parameter in function,`cleanText()`) in `for punc in WordSet.replacePuncs:`

BUT we can't do this with `self.cleanText(text)` because that would actually pass the instance into this function, `cleanText()`. We have to use `WordSet.`

### Decorators 

A **decorator** in Python is an annotation, or description for your function definition.

It defines some special attribute or information about the function, so that Python knows how to handle it.

```python
class WordSet:
    replacePuncs = ['!', '.', ',', '\'']
    def __init__(self):
        self.words = set()
        
    def addText(self, text):
        text = self.cleanText(text) #can use `self`
        for word in text.split():
            self.words.add(word)
            
    @staticmethod #this is a static method decorator
    def cleanText(text): #Explicitly told Python cleanText is a static method, so `self` should not be passed in as an argument. 
        # chaining functions
        for punc in WordSet.replacePuncs:
            text = text.replace(punc, '')
        return text.lower()
    
        
wordSet = WordSet()

wordSet.addText('Hi, I\'m Ryan! Here is a sentence I want to add!')
wordSet.addText('Here is another sentence I want to add.')

print(wordSet.words)
```
