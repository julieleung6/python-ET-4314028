## Handling Exceptions

### Try / Except

```python
def causeError():
    try:
        return 1/0
    except Exception as e:
        return e
    
causeError()
```

Here, we are catching this exception and then just returning it. So we don't get a stack trace, but we do see that this `ZeroDivisionError` instance is returned.

```python
# If we dont care about getting the specific instance of the exception, and just want to print something
def causeError():
    try:
        return 1/0
    except Exception:
        print('There was some sort of error!')
    
causeError()
```

We don't have to have the `as e` in order to catch an exception. "There was some sort of error!" and that just prints out.

### Finally

```python
def causeError():
    try:
        return 1/0
    except Exception:
        print('There was some sort of error!')
    finally:
        print('This will always execute!')
    
causeError()
```

**Finally** statements can be useful because they will *always* execute no matter what happens inside this `try` block.

```python
def causeError():
    try:
        return 1/0
    finally:
        print('This will always execute!')
    
causeError()
```

You don't even need any `except` statements. This error is thrown but this still printed out the message 'This will always execute!'

```python
def causeError():
    try:
        return 1/1
    finally:
        print('This will always execute!')
    
causeError()
```

Even if no exception is raised at all, that still executes.

I often use `finally` statements when timing how long a function takes to execute.

```python
def causeError():
    start = time.time()
    try:
        time.sleep(0.5) #this just pauses execution for some number of seconds. Here we are pausing for half a second
        return 1/1
    except Exception:
        print('There was some sort of error!')
    finally:
        print(f'Function took {time.time() - start} seconds to execute')
    
causeError()
```

```python
def causeError():
    start = time.time()
    try:
        time.sleep(0.5)
        return 1/0
    except Exception:
        print('There was some sort of error!')
    finally:
        print(f'Function took {time.time() - start} seconds to execute')
    
causeError()
```

This `try-finally` pattern keeps your code clean and compact and lets you do any needed cleanup or logging after a statement completes no matter what happens inside the `try` block.

### Catching Exceptions by Type

```python
def causeError():
    try:
        1/0

    except TypeError:
        print('There was a type error!')
    except ZeroDivisionError:
        print('There was a zero division error!')
    except Exception:
        print('There was some sort of error!')

    
causeError()
```

```python
def causeError():
    try:
        return 1 + 'a'

    except TypeError:
        print('There was a type error!')
    except ZeroDivisionError:
        print('There was a zero division error!')
    except Exception:
        print('There was some sort of error!')

    
causeError()
```

The order of these `except` statements **does matter** here.

So Python will try the first one, and it doesn't bother with the other ones if first one is true.

```python
def causeError():
    try:
        return 1 + 'a'
        
    except Exception: #moving this general Exception up here
        print('There was some sort of error!')
    except TypeError:
        print('There was a type error!')
    except ZeroDivisionError:
        print('There was a zero division error!')

    
causeError()
```

If I move this general exception up, this is the class from which all of these extend, there was some of error.

**Note**: You always want your most general `Exception` statements at the bottom. And the most specific ones up at the top.

### Custom Decorators

They can:
- Modify the arguments of a function
- Set up or tear down before and after a function runs.
- Modify the return value of a function.

Sometimes you are doing really involved exception handling and catching.

For instance, a lot of HTTP request response handling where there are a lot of different types of HTTP errors and you'll have a lot of `except` statements all in a row. And then you'll just copy and paste all of these different blocks into many different functions. And in this situation, it can be really handy to move this trying and catching into a *single* function. You can use a custom decorator.

```python
def handleException(func):
    def wrapper():
        try:
            func()
        except TypeError:
            print('There was a type error!')
        except ZeroDivisionError:
            print('There was a zero division error!')
        except Exception:
            print('There was some sort of error!')
    return wrapper

@handleException
def causeError():
    return 1/0

causeError()
```

You can see that this `handleException` was used to `except` those various exceptions that this could throw.

### Raising Exceptions

```python
@handleException # reusing our handleException decorator for another function
def raiseError(n):
    if n == 0: #this function accepts any input except the number, zero.
        raise Exception() #this raise statement will raises/throws this new exception that I'm creating when its reached.
    print(n) #
    
raiseError()
```

Don't need an `else` statement because once the `Exception` is raised here, this execution will just halt. It will throw this `Exception`, and then the `print(n)` will never be reached.

```python
def handleException(func): #raiseError is getting passed in as func
    def wrapper(*args): #modifying it by using variable args
        try:
            #func() #but when we call it, there aren't any arguments.
            func(*args) #passing args here
        except TypeError:
            print('There was a type error!')
        except ZeroDivisionError:
            print('There was a zero division error!')
        except Exception:
            print('There was some sort of error!')
    return wrapper

@handleException
def causeError():
    return 1/0

causeError()
```

```python
@handleException #trying to use this handler on a function, raiseError() that takes arguments.
def raiseError(n): #even though, our function has an argument.
    if n == 0:
        raise Exception()
    print(n) #
    
raiseError(1) #so now, we get an error printed if the argument we pass in is zero, otherwise it just prints out the value
```

Writing functions to raise exceptions is powerful when you combine it with custom exceptions.
