## While loops

**Note**: This notebook has been modified from what is shown in the video. In order to prevent bugs when the current time ends in 58 or 59 seconds, the following line:

<code>wait_until = datetime.now().second + 2</code>

has been changed to 

<code>wait_until = (datetime.now().second + 2) % 60</code>

where it appears in the code samples

```python
datetime.now().second
```

```python
datetime.now().second + 2
```

```python
# we want to get two seconds into the future, just add 2 BUT have to be careful with this!!!
# For example, we are currently at 59 seconds, it's at one second into the current minute, so we need to use modulus 60 to reflect that.
wait_until = (datetime.now().second + 2) % 60

# This while loop waits 2 seconds and then prints out a message
while datetime.now().second != wait_until:
    print('Still waiting!')
    
print(f'We are at {wait_until} seconds!')
```

```python
wait_until = datetime.now().second + 2

while datetime.now().second != wait_until:
    1 + 1 # need to have something indented under the loop. Can do some useless calculation but that is a waste of computing power!!
    
print(f'We are at {wait_until} seconds!')
```

### Pass

- Need to have something indented under the *while loop*
- `pass` statement basically says, there's nothing to see here move along but it does critically preserve our indenting.
- Can be used as a placeholder if you are writing a function or a class definition, but don't want to fill it in yet.

```python
wait_until = (datetime.now().second + 2) % 60

while datetime.now().second != wait_until:
    pass
    
print(f'We are at {wait_until} seconds!')
```

### Break

- `break` statement basically just breaks out of the current loop that it's in.
  
**Note:** the `break` statement only breaks out of the *first loop* that it's in.

```python
wait_until = (datetime.now().second + 2) % 60

while True: # be careful when using "while True"!!!
    if datetime.now().second == wait_until: 
        print(f'We are at {wait_until} seconds!')
        break
```

```python
wait_until = (datetime.now().second + 2) % 60

while True:
    while datetime.now().second == wait_until: #This will run FOREVER!!
        print(f'We are at {wait_until} seconds!')
        break
```

### Continue

- `continue` statement skips any lines that follow it inside of the *while loop*.
- Usually you see a `continue` statement used inside an `if` statement to *prevent the rest of the code in the loop from being executed* but only if some condition is met.

```python
wait_until = (datetime.now().second + 2) % 60

while datetime.now().second != wait_until:
    continue #this skips everything else that follows it inside this while loop
    print('Still waiting!') #this line never prints
    
    
print(f'We are at {wait_until} seconds!')
```

```python
wait_until = (datetime.now().second + 2) % 60

while True:
    if datetime.now().second < wait_until:
        continue #when this statement is hit, it skips everything else in the loop, which is just the break statement.
        #as soon as the if statement is no longer true, then continue is no longer hit. It reaches the break statement
    break #breaking it outside of this while loop
    
print(f'We are at {wait_until} seconds!') #then it prints this
```

`continue` and `break` statements are not technically logically necessary, but they can help arrange and write code more clearly for readability.
